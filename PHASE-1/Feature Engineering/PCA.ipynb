```python
import pandas as pd
import numpy as np

np.random.seed(42)

df = pd.DataFrame({
    "Age": np.random.randint(18, 65, 100),
    "Income": np.random.randint(20000, 150000, 100),
    "Experience": np.random.randint(0, 35, 100),
    "CreditScore": np.random.randint(300, 850, 100),
    "LoanAmount": np.random.randint(5000, 500000, 100),
    "AnnualSpend": np.random.randint(1000, 100000, 100),
    "Savings": np.random.randint(500, 200000, 100),
    "Debt": np.random.randint(0, 300000, 100),
    "Transactions": np.random.randint(10, 500, 100),
    "Target": np.random.randint(0, 2, 100)
})
print(df.shape)
```

```python
from sklearn.decomposition import PCA # Importing of PCA
```

```python
from sklearn.preprocessing import StandardScaler # Importing of Scaler
```

```python
scaler = StandardScaler() # Declaration of Scaler
x = df.drop(columns=['Target'])
x_scaled = scaler.fit_transform(x)
```

```python
pca = PCA(n_components=0.95) # Declaration of PCA
x_reduced = pca.fit_transform(x_scaled)
#print(x_reduced)
```

```python
pca_df = pd.DataFrame(
    x_reduced,
    columns=[f"PC{i+1}" for i in range(x_reduced.shape[1])]
)

print(pca_df.shape)
```
