```python
import pandas as pd
import numpy as np

np.random.seed(42)

df = pd.DataFrame({
    "Age": np.random.randint(18, 65, 100),
    "Income": np.random.randint(20000, 150000, 100),
    "Experience": np.random.randint(0, 35, 100),
    "CreditScore": np.random.randint(300, 850, 100),
    "LoanAmount": np.random.randint(5000, 500000, 100),
    "AnnualSpend": np.random.randint(1000, 100000, 100),
    "Savings": np.random.randint(500, 200000, 100),
    "Debt": np.random.randint(0, 300000, 100),
    "Transactions": np.random.randint(10, 500, 100),
    "Target": np.random.randint(0, 2, 100)
})
print(df.shape)
```

```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler # Importing the scaling functions
```

```python
x = df.drop(columns=['Target'])
y = df['Target']
```

```python
standard_scaler = StandardScaler() # Declaration of Standard Scaler
x_scaled_1 = standard_scaler.fit_transform(x) # Scaling and transforming using scaler
```

```python
standard_scaler = pd.DataFrame(
    x_scaled_1,
    columns=[f"PC{i+1}" for i in range(x_scaled_1.shape[1])]
)

print(standard_scaler.shape)
```

```python
minmax_scaler = MinMaxScaler() # Declaration of MinMax Scaler
x_scaled_2 = minmax_scaler.fit_transform(x) # Scaling and transforming using scaler
```

```python
minmax_sacler_df = pd.DataFrame(
    x_scaled_2,
    columns=[f"PC{i+1}" for i in range(x_scaled_2.shape[1])]
)

print(minmax_sacler_df.shape)
```
