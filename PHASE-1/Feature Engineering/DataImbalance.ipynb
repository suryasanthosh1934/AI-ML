```python
print(df['Target'].value_counts())
```

```python
from imblearn.over_sampling import SMOTE  # Importing SMOTE

smote = SMOTE( k_neighbors=1,random_state=42)  # Declaration of SMOTE

X_resampled, y_resampled = smote.fit_resample(
    df.drop('Target', axis=1),
    df['Target']
)


print(y_resampled.value_counts())
```

```python
from imblearn.over_sampling import ADASYN  # Importing ADASYN

adasyn = ADASYN(
    n_neighbors=1,
    random_state=42
)  # Declaration of ADASYN

X_resampled_2, y_resampled_2 = adasyn.fit_resample(
    df.drop('Target', axis=1),
    df['Target']
)

print(y_resampled_2.value_counts())
```
