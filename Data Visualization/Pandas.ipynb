```python
# Question change the education of the girl named neha to MBA
for i in range(len(df)) :
  if df["names"][i]=="neha":
    df["education"][i]= "MBA"

print(df)
```

```python
# Before training the model with the data should be duplicated and null values should be removed
```

```python
# case 1 (null values)
df = df.dropna() # Used to remove the null values
print(df)
```

```python
# case 1 (null values)
df = df.fillna(1) # you can fill the null values with certain values like mean , median , mode etc
print(df)
```

```python
#case 2(duplicates)
df = df.drop_duplicates() # used to remove the duplicates
print(df)
```
