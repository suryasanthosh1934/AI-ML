```python
import numpy as np # import numpy package using alias name
```

```python
arr = np.array([1,2,3]) # Declaration of an array using numpy
print(arr)
```

```python
list_user = list((input("enter the number : "))) # Declaration of numpy
arr_user = np.array(list_user) # Taking the input from the user
print(arr_user)
```

```python
print(type(arr_user)) # used to find the datatype of the array
print(arr_user.ndim) # To find the dimension of the array
print(arr_user.shape) # To find the shape of the array
```

```python
arr_ndim = np.array([[1,2,3],[4,5,6]]) # Declartion of 2D array
print(arr_ndim)
```

```python
arr_1 = np.ones((2,3)) # Declartion of an array of dimension (2,3) with each element "1"
print(arr_1)
```

```python
arr_1 = np.zeros((2,3)) # Declartion of an array of dimension (2,3) with each element "0"
print(arr_1)
```

```python
arr_unity = np.eye(3) # Declartion of unit matrix of dimension 3*3
print(arr_unity)
```

```python
arr_loop = np.arange(0,10,2)
print(arr_loop)
```

```python
arr_loop_ap = np.linspace(0,10,2)
print(arr_loop_ap)
```

```python
arr_loop_gp = np.logspace(0,10,2)
print(arr_loop_gp)
```

```python
arr_random = np.random.rand(2,3) # Declartion of an array of dimension (2,3) with random elements
print(arr_random)
```

```python
arr_random = np.random.randn(2,3) # Declartion of an array of dimension (2,3) with random elements which form standard normal deviation mean = 0 and SD = 1
print(arr_random)
```

```python
arr_random = np.random.normal(loc = 5,scale = 2,size = (2,3)) # Declartion of an array of dimension (2,3) with random elements which form standard normal deviation mean = 5 and SD = 2
print(arr_random)
```
